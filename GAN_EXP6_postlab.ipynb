```python
# ============================================================
# VANILLA GAN USING FASHION-MNIST - POST LAB
# Aim: Generate synthetic clothing images using GAN
#
# Post-Lab:
# 1. Visualize generator outputs at different epochs
# 2. Experiment with deeper architectures
# 3. Automatically save generated images during training
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import tensorflow as tf
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt
import os


# ============================================================
# 2. LOAD FASHION-MNIST DATASET
# ============================================================

(x_train, _), (_, _) = tf.keras.datasets.fashion_mnist.load_data()

print("Training images shape:", x_train.shape)


# ============================================================
# 3. NORMALIZE DATA
# ============================================================

# Convert pixel values from 0-255 to -1 to 1

x_train = x_train.astype("float32")

x_train = (x_train - 127.5) / 127.5


# ============================================================
# 4. FLATTEN IMAGES
# ============================================================

# Convert 28 x 28 images into 784 values

x_train = x_train.reshape(-1, 784)

print("After reshaping:", x_train.shape)


# ============================================================
# 5. CREATE DATASET BATCHES
# ============================================================

BUFFER_SIZE = 60000
BATCH_SIZE = 256

dataset = (
    tf.data.Dataset
    .from_tensor_slices(x_train)
    .shuffle(BUFFER_SIZE)
    .batch(BATCH_SIZE)
)


# ============================================================
# 6. BUILD DEEPER GENERATOR
# ============================================================

# Generator takes 100 random values
# and generates a 784-value image

generator = tf.keras.Sequential([

    layers.Input(shape=(100,)),

    layers.Dense(256),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(512),
    layers.LeakyReLU(negative_slope=0.2),

    # Additional layer for deeper architecture
    layers.Dense(1024),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(784, activation="tanh")
])


print("\n================ GENERATOR ================")
generator.summary()


# ============================================================
# 7. BUILD DEEPER DISCRIMINATOR
# ============================================================

# Discriminator checks whether image is real or fake

discriminator = tf.keras.Sequential([

    layers.Input(shape=(784,)),

    layers.Dense(1024),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(512),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(256),
    layers.LeakyReLU(negative_slope=0.2),

    layers.Dense(1, activation="sigmoid")
])


print("\n============== DISCRIMINATOR ==============")
discriminator.summary()


# ============================================================
# 8. LOSS FUNCTION AND OPTIMIZERS
# ============================================================

loss_function = tf.keras.losses.BinaryCrossentropy()

generator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002
)

discriminator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002
)


# ============================================================
# 9. CREATE FOLDER TO SAVE GENERATED IMAGES
# ============================================================

os.makedirs("generated_images", exist_ok=True)

print("\nGenerated images will be saved in:")
print("generated_images/")


# ============================================================
# 10. TRAINING FUNCTION
# ============================================================

def train_step(real_images):

    # Get current batch size
    batch_size = tf.shape(real_images)[0]

    # Generate random noise
    noise = tf.random.normal([batch_size, 100])


    # --------------------------------------------------------
    # TRAIN DISCRIMINATOR
    # --------------------------------------------------------

    with tf.GradientTape() as disc_tape:

        # Generate fake images
        fake_images = generator(
            noise,
            training=True
        )

        # Discriminator prediction for real images
        real_output = discriminator(
            real_images,
            training=True
        )

        # Discriminator prediction for fake images
        fake_output = discriminator(
            fake_images,
            training=True
        )

        # Real image loss
        real_loss = loss_function(
            tf.ones_like(real_output),
            real_output
        )

        # Fake image loss
        fake_loss = loss_function(
            tf.zeros_like(fake_output),
            fake_output
        )

        # Total discriminator loss
        discriminator_loss = real_loss + fake_loss


    # Calculate discriminator gradients

    discriminator_gradients = disc_tape.gradient(
        discriminator_loss,
        discriminator.trainable_variables
    )


    # Apply discriminator gradients

    discriminator_optimizer.apply_gradients(
        zip(
            discriminator_gradients,
            discriminator.trainable_variables
        )
    )


    # --------------------------------------------------------
    # TRAIN GENERATOR
    # --------------------------------------------------------

    # Generate new random noise

    noise = tf.random.normal(
        [batch_size, 100]
    )


    with tf.GradientTape() as gen_tape:

        # Generate fake images

        fake_images = generator(
            noise,
            training=True
        )

        # Discriminator prediction

        fake_output = discriminator(
            fake_images,
            training=True
        )

        # Generator wants fake images
        # to be classified as real

        generator_loss = loss_function(
            tf.ones_like(fake_output),
            fake_output
        )


    # Calculate generator gradients

    generator_gradients = gen_tape.gradient(
        generator_loss,
        generator.trainable_variables
    )


    # Apply generator gradients

    generator_optimizer.apply_gradients(
        zip(
            generator_gradients,
            generator.trainable_variables
        )
    )


    return generator_loss, discriminator_loss


# ============================================================
# 11. FUNCTION TO GENERATE AND DISPLAY IMAGES
# ============================================================

def generate_and_show_images(epoch):

    # Generate 16 random noise vectors

    noise = tf.random.normal(
        [16, 100]
    )


    # Generate images

    generated_images = generator(
        noise,
        training=False
    )


    # Convert from [-1, 1] to [0, 1]

    generated_images = (
        generated_images + 1
    ) / 2.0


    # Create figure

    plt.figure(figsize=(6, 6))


    # Display 16 images

    for i in range(16):

        plt.subplot(4, 4, i + 1)

        image = generated_images[
            i
        ].numpy().reshape(28, 28)

        plt.imshow(
            image,
            cmap="gray"
        )

        plt.axis("off")


    plt.suptitle(
        f"Generated Images - Epoch {epoch}"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 12. FUNCTION TO SAVE GENERATED IMAGES
# ============================================================

def save_generated_images(epoch):

    # Generate random noise

    noise = tf.random.normal(
        [16, 100]
    )


    # Generate images

    generated_images = generator(
        noise,
        training=False
    )


    # Convert from [-1, 1] to [0, 1]

    generated_images = (
        generated_images + 1
    ) / 2.0


    # Create figure

    plt.figure(figsize=(6, 6))


    # Display images

    for i in range(16):

        plt.subplot(4, 4, i + 1)

        image = generated_images[
            i
        ].numpy().reshape(28, 28)

        plt.imshow(
            image,
            cmap="gray"
        )

        plt.axis("off")


    plt.suptitle(
        f"Epoch {epoch}"
    )

    plt.tight_layout()


    # File name

    filename = (
        f"generated_images/epoch_{epoch}.png"
    )


    # Save image

    plt.savefig(
        filename,
        dpi=150,
        bbox_inches="tight"
    )


    # Close figure

    plt.close()


    print(
        "Generated images saved:",
        filename
    )


# ============================================================
# 13. TRAIN THE GAN
# ============================================================

EPOCHS = 20


print("\n================ TRAINING ================\n")


for epoch in range(EPOCHS):

    gen_loss = 0
    disc_loss = 0


    # Train using all batches

    for batch in dataset:

        gen_loss, disc_loss = train_step(
            batch
        )


    # Print losses

    print(
        f"Epoch {epoch + 1}/{EPOCHS} | "
        f"Generator Loss: {gen_loss.numpy():.4f} | "
        f"Discriminator Loss: {disc_loss.numpy():.4f}"
    )


    # --------------------------------------------------------
    # Save and display images every 5 epochs
    # --------------------------------------------------------

    if (epoch + 1) % 5 == 0:

        # Save generated images

        save_generated_images(
            epoch + 1
        )

        # Display generated images

        generate_and_show_images(
            epoch + 1
        )


print("\nGAN Training Completed!")


# ============================================================
# 14. GENERATE FINAL SYNTHETIC IMAGES
# ============================================================

noise = tf.random.normal(
    [16, 100]
)


generated_images = generator(
    noise,
    training=False
)


# Convert [-1,1] to [0,1]

generated_images = (
    generated_images + 1
) / 2.0


# ============================================================
# 15. DISPLAY FINAL GENERATED IMAGES
# ============================================================

plt.figure(figsize=(7, 7))


for i in range(16):

    plt.subplot(4, 4, i + 1)

    image = generated_images[
        i
    ].numpy().reshape(28, 28)

    plt.imshow(
        image,
        cmap="gray"
    )

    plt.axis("off")


plt.suptitle(
    "Final Synthetic Fashion Images"
)

plt.tight_layout()

plt.show()


# ============================================================
# 16. POST-LAB RESULT
# ============================================================

print("\n==========================================")
print("POST-LAB EXPERIMENT COMPLETED")
print("==========================================")

print("1. Generator outputs visualized at different epochs.")
print("2. Deeper Generator and Discriminator were used.")
print("3. Generated images were automatically saved.")
print("4. Final synthetic clothing images were displayed.")
print("==========================================")
```
