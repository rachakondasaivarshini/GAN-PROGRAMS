```python
# ============================================================
# POST-LAB: TRANSFER LEARNING AND FINE-TUNING
#
# Dataset : Oxford-IIIT Pet Dataset
# Models  : MobileNetV2 and ResNet50
#
# Tasks:
# 1. Fine-tune MobileNetV2
# 2. Add Data Augmentation using ImageDataGenerator
# 3. Replace MobileNetV2 with ResNet50
# 4. Compare performance
# 5. Confusion Matrix
# 6. Visualize Misclassified Images
# 7. Save and Load Model
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import tarfile
import urllib.request
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2, ResNet50
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as mobilenet_preprocess
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet_preprocess

from tensorflow.keras import layers, models
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split

print("TensorFlow Version:", tf.__version__)


# ============================================================
# 2. DOWNLOAD OXFORD-IIIT PET DATASET
# ============================================================

url = "https://www.robots.ox.ac.uk/~vgg/data/pets/data/images.tar.gz"

archive_path = "/content/images.tar.gz"
extract_path = "/content/oxford_pets"
image_dir = "/content/oxford_pets/images"

os.makedirs(extract_path, exist_ok=True)

if not os.path.exists(archive_path):

    print("\nDownloading Oxford-IIIT Pet Dataset...")

    urllib.request.urlretrieve(
        url,
        archive_path
    )

    print("Download completed.")

else:

    print("\nDataset archive already downloaded.")


# ============================================================
# 3. EXTRACT DATASET
# ============================================================

if not os.path.exists(image_dir):

    print("\nExtracting dataset...")

    with tarfile.open(
        archive_path,
        "r:gz"
    ) as tar:

        tar.extractall(extract_path)

    print("Extraction completed.")

else:

    print("\nImages already extracted.")


# ============================================================
# 4. GET IMAGE FILES AND BREED LABELS
# ============================================================

image_files = [
    f for f in os.listdir(image_dir)
    if f.lower().endswith(".jpg")
]


def get_breed(filename):

    name = os.path.splitext(filename)[0]

    return name.rsplit("_", 1)[0]


class_names = sorted(
    list(
        set(
            get_breed(f)
            for f in image_files
        )
    )
)

num_classes = len(class_names)

print("\nTotal Images:", len(image_files))
print("Number of Classes:", num_classes)


# ============================================================
# 5. CREATE DATAFRAME
# ============================================================

data = []

for filename in image_files:

    breed = get_breed(filename)

    data.append(
        [
            os.path.join(image_dir, filename),
            breed
        ]
    )


df = pd.DataFrame(
    data,
    columns=["filename", "class"]
)

print("\nDataset DataFrame:")
print(df.head())


# ============================================================
# 6. LIMIT DATASET
# ============================================================

# Set to None to use all images.
# 3000 is faster for demonstration in Colab.

MAX_IMAGES = 3000

if MAX_IMAGES is not None and len(df) > MAX_IMAGES:

    df, _ = train_test_split(
        df,
        train_size=MAX_IMAGES,
        stratify=df["class"],
        random_state=42
    )

    df = df.reset_index(drop=True)


print("\nImages used for experiment:", len(df))


# ============================================================
# 7. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["class"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["class"],
    random_state=42
)

print("\nDataset Split")
print("----------------------------")
print("Training Images:", len(train_df))
print("Validation Images:", len(val_df))
print("Testing Images:", len(test_df))


# ============================================================
# 8. PARAMETERS
# ============================================================

IMG_SIZE = 160
BATCH_SIZE = 32


# ============================================================
# 9. DATA AUGMENTATION
# ============================================================

train_datagen = ImageDataGenerator(
    preprocessing_function=mobilenet_preprocess,

    rotation_range=20,

    width_shift_range=0.2,

    height_shift_range=0.2,

    shear_range=0.2,

    zoom_range=0.2,

    horizontal_flip=True,

    fill_mode="nearest"
)


val_test_datagen = ImageDataGenerator(
    preprocessing_function=mobilenet_preprocess
)


# ============================================================
# 10. CREATE MOBILE NET GENERATORS
# ============================================================

train_generator = train_datagen.flow_from_dataframe(

    dataframe=train_df,

    x_col="filename",

    y_col="class",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=True,

    seed=42
)


val_generator = val_test_datagen.flow_from_dataframe(

    dataframe=val_df,

    x_col="filename",

    y_col="class",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=False
)


test_generator = val_test_datagen.flow_from_dataframe(

    dataframe=test_df,

    x_col="filename",

    y_col="class",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=False
)


# ============================================================
# 11. CREATE MOBILE NET MODEL
# ============================================================

print("\nLoading MobileNetV2...")

mobilenet_base = MobileNetV2(

    input_shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    ),

    include_top=False,

    weights="imagenet"
)

# Freeze all base layers initially

mobilenet_base.trainable = False


mobilenet_model = models.Sequential([

    mobilenet_base,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.3),

    layers.Dense(
        num_classes,
        activation="softmax"
    )
])


# ============================================================
# 12. COMPILE MOBILE NET MODEL
# ============================================================

mobilenet_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


print("\nMobileNetV2 Model:")
mobilenet_model.summary()


# ============================================================
# 13. TRAIN MOBILE NET
# ============================================================

print("\n==============================")
print("Training MobileNetV2")
print("==============================")


history_mobilenet = mobilenet_model.fit(

    train_generator,

    validation_data=val_generator,

    epochs=3
)


# ============================================================
# 14. FINE-TUNE MOBILE NET
# ============================================================

print("\n==============================")
print("Fine-Tuning MobileNetV2")
print("==============================")


# Unfreeze the base model

mobilenet_base.trainable = True


# Freeze all layers except the last 20

for layer in mobilenet_base.layers[:-20]:

    layer.trainable = False


# Recompile with smaller learning rate

mobilenet_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.00001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


# Fine-tune

history_finetune = mobilenet_model.fit(

    train_generator,

    validation_data=val_generator,

    epochs=2
)


# ============================================================
# 15. EVALUATE MOBILE NET
# ============================================================

mobilenet_loss, mobilenet_accuracy = (
    mobilenet_model.evaluate(
        test_generator,
        verbose=1
    )
)


print("\n==============================")
print("MobileNetV2 PERFORMANCE")
print("==============================")

print(
    "Test Loss:",
    round(mobilenet_loss, 4)
)

print(
    "Test Accuracy:",
    round(
        mobilenet_accuracy * 100,
        2
    ),
    "%"
)


# ============================================================
# 16. CREATE RESNET50 DATA GENERATORS
# ============================================================

resnet_train_datagen = ImageDataGenerator(

    preprocessing_function=resnet_preprocess,

    rotation_range=20,

    width_shift_range=0.2,

    height_shift_range=0.2,

    shear_range=0.2,

    zoom_range=0.2,

    horizontal_flip=True,

    fill_mode="nearest"
)


resnet_val_test_datagen = ImageDataGenerator(

    preprocessing_function=resnet_preprocess
)


resnet_train_generator = (
    resnet_train_datagen.flow_from_dataframe(

        dataframe=train_df,

        x_col="filename",

        y_col="class",

        target_size=(IMG_SIZE, IMG_SIZE),

        batch_size=BATCH_SIZE,

        class_mode="categorical",

        shuffle=True,

        seed=42
    )
)


resnet_val_generator = (
    resnet_val_test_datagen.flow_from_dataframe(

        dataframe=val_df,

        x_col="filename",

        y_col="class",

        target_size=(IMG_SIZE, IMG_SIZE),

        batch_size=BATCH_SIZE,

        class_mode="categorical",

        shuffle=False
    )
)


resnet_test_generator = (
    resnet_val_test_datagen.flow_from_dataframe(

        dataframe=test_df,

        x_col="filename",

        y_col="class",

        target_size=(IMG_SIZE, IMG_SIZE),

        batch_size=BATCH_SIZE,

        class_mode="categorical",

        shuffle=False
    )
)


# ============================================================
# 17. LOAD RESNET50
# ============================================================

print("\nLoading ResNet50...")


resnet_base = ResNet50(

    input_shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    ),

    include_top=False,

    weights="imagenet"
)


# Freeze pre-trained layers

resnet_base.trainable = False


# ============================================================
# 18. CREATE RESNET50 MODEL
# ============================================================

resnet_model = models.Sequential([

    resnet_base,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.3),

    layers.Dense(
        num_classes,
        activation="softmax"
    )
])


# ============================================================
# 19. COMPILE RESNET50
# ============================================================

resnet_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


print("\nResNet50 Model:")
resnet_model.summary()


# ============================================================
# 20. TRAIN RESNET50
# ============================================================

print("\n==============================")
print("Training ResNet50")
print("==============================")


history_resnet = resnet_model.fit(

    resnet_train_generator,

    validation_data=resnet_val_generator,

    epochs=3
)


# ============================================================
# 21. FINE-TUNE RESNET50
# ============================================================

print("\n==============================")
print("Fine-Tuning ResNet50")
print("==============================")


resnet_base.trainable = True


# Freeze all except top 20 layers

for layer in resnet_base.layers[:-20]:

    layer.trainable = False


resnet_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.00001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


history_resnet_finetune = resnet_model.fit(

    resnet_train_generator,

    validation_data=resnet_val_generator,

    epochs=2
)


# ============================================================
# 22. EVALUATE RESNET50
# ============================================================

resnet_loss, resnet_accuracy = (
    resnet_model.evaluate(
        resnet_test_generator,
        verbose=1
    )
)


print("\n==============================")
print("ResNet50 PERFORMANCE")
print("==============================")

print(
    "Test Loss:",
    round(resnet_loss, 4)
)

print(
    "Test Accuracy:",
    round(
        resnet_accuracy * 100,
        2
    ),
    "%"
)


# ============================================================
# 23. COMPARE MOBILE NET AND RESNET50
# ============================================================

print("\n==============================")
print("MODEL COMPARISON")
print("==============================")


print(
    "MobileNetV2 Accuracy:",
    round(
        mobilenet_accuracy * 100,
        2
    ),
    "%"
)


print(
    "ResNet50 Accuracy:",
    round(
        resnet_accuracy * 100,
        2
    ),
    "%"
)


if mobilenet_accuracy > resnet_accuracy:

    print(
        "\nBetter Model: MobileNetV2"
    )

elif resnet_accuracy > mobilenet_accuracy:

    print(
        "\nBetter Model: ResNet50"
    )

else:

    print(
        "\nBoth models have the same accuracy."
    )


# ============================================================
# 24. PERFORMANCE COMPARISON GRAPH
# ============================================================

models_names = [
    "MobileNetV2",
    "ResNet50"
]

accuracies = [
    mobilenet_accuracy * 100,
    resnet_accuracy * 100
]


plt.figure(
    figsize=(7, 5)
)

plt.bar(
    models_names,
    accuracies
)

plt.ylabel("Test Accuracy (%)")

plt.xlabel("Model")

plt.title(
    "MobileNetV2 vs ResNet50"
)

plt.ylim(0, 100)

for i, value in enumerate(accuracies):

    plt.text(
        i,
        value + 1,
        f"{value:.2f}%",
        ha="center"
    )

plt.show()


# ============================================================
# 25. CONFUSION MATRIX - MOBILE NET
# ============================================================

print("\nGenerating MobileNetV2 Confusion Matrix...")


test_generator.reset()

mobilenet_predictions = (
    mobilenet_model.predict(
        test_generator,
        verbose=1
    )
)


mobilenet_pred_classes = np.argmax(
    mobilenet_predictions,
    axis=1
)


true_classes = test_generator.classes


cm_mobilenet = confusion_matrix(

    true_classes,

    mobilenet_pred_classes
)


plt.figure(
    figsize=(16, 16)
)

disp = ConfusionMatrixDisplay(

    confusion_matrix=cm_mobilenet,

    display_labels=class_names
)

disp.plot(

    xticks_rotation=90,

    ax=plt.gca(),

    cmap="Blues",

    colorbar=False
)

plt.title(
    "MobileNetV2 Confusion Matrix"
)

plt.show()


# ============================================================
# 26. FIND MISCLASSIFIED IMAGES
# ============================================================

misclassified_indices = np.where(

    mobilenet_pred_classes != true_classes

)[0]


print(
    "\nTotal Misclassified Images:",
    len(misclassified_indices)
)


# ============================================================
# 27. VISUALIZE MISCLASSIFIED IMAGES
# ============================================================

print(
    "\nDisplaying Misclassified Images..."
)


num_display = min(
    9,
    len(misclassified_indices)
)


plt.figure(
    figsize=(12, 12)
)


for i in range(num_display):

    index = misclassified_indices[i]

    image_path = test_df.iloc[index]["filename"]

    actual_class = class_names[
        true_classes[index]
    ]

    predicted_class = class_names[
        mobilenet_pred_classes[index]
    ]


    image = tf.keras.utils.load_img(

        image_path,

        target_size=(
            IMG_SIZE,
            IMG_SIZE
        )
    )


    plt.subplot(
        3,
        3,
        i + 1
    )

    plt.imshow(image)

    plt.title(
        "Actual: "
        + actual_class.replace("_", " ")
        + "\nPredicted: "
        + predicted_class.replace("_", " "),
        fontsize=9
    )

    plt.axis("off")


plt.tight_layout()

plt.show()


# ============================================================
# 28. SAVE MOBILENETV2 MODEL
# ============================================================

mobilenet_save_path = (
    "/content/mobilenetv2_pet_model.keras"
)


mobilenet_model.save(
    mobilenet_save_path
)


print(
    "\nMobileNetV2 model saved at:"
)

print(
    mobilenet_save_path
)


# ============================================================
# 29. LOAD SAVED MOBILENETV2 MODEL
# ============================================================

loaded_model = tf.keras.models.load_model(

    mobilenet_save_path
)


print(
    "\nSaved MobileNetV2 model loaded successfully!"
)


# ============================================================
# 30. EVALUATE LOADED MODEL
# ============================================================

loaded_loss, loaded_accuracy = (

    loaded_model.evaluate(
        test_generator,
        verbose=1
    )
)


print("\n==============================")
print("LOADED MODEL PERFORMANCE")
print("==============================")


print(
    "Loaded Model Loss:",
    round(
        loaded_loss,
        4
    )
)


print(
    "Loaded Model Accuracy:",
    round(
        loaded_accuracy * 100,
        2
    ),
    "%"
)


# ============================================================
# 31. FINAL RESULT
# ============================================================

print("\n================================================")
print("              POST-LAB COMPLETED")
print("================================================")

print(
    "MobileNetV2 Test Accuracy:",
    round(
        mobilenet_accuracy * 100,
        2
    ),
    "%"
)

print(
    "ResNet50 Test Accuracy:",
    round(
        resnet_accuracy * 100,
        2
    ),
    "%"
)

print(
    "Loaded Model Accuracy:",
    round(
        loaded_accuracy * 100,
        2
    ),
    "%"
)

print(
    "Misclassified Images:",
    len(misclassified_indices)
)

print(
    "\nModel saved successfully."
)

print(
    "Model loaded successfully."
)

print(
    "\nPOST-LAB PROGRAM COMPLETED SUCCESSFULLY!"
)

print("================================================")
```
