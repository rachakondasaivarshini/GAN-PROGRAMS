```python
# Install dataset library
!pip -q install datasets

# Import libraries
import numpy as np
import tensorflow as tf
from datasets import load_dataset
from tensorflow.keras import layers, models

# Load GTSRB dataset
data = load_dataset("tanganke/gtsrb")

print("Dataset Loaded Successfully!")

# Select data
train_data = data["train"].select(range(5000))
test_data = data["test"].select(range(1000))

print("Training Images:", len(train_data))
print("Testing Images :", len(test_data))

# Preprocess images
def prepare_data(dataset):
    images = []
    labels = []

    for item in dataset:
        img = item["image"].convert("RGB")
        img = img.resize((32, 32))
        img = np.array(img) / 255.0

        images.append(img)
        labels.append(item["label"])

    return np.array(images, dtype="float32"), np.array(labels)

# Prepare data
X_train, y_train = prepare_data(train_data)
X_test, y_test = prepare_data(test_data)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape :", X_test.shape)

# Create CNN
model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),

    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(128, activation="relu"),

    layers.Dense(43, activation="softmax")
])

# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Show model
model.summary()

# Train
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)

# Test
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)

print("\nTest Accuracy:", accuracy * 100, "%")

# Predict one image
test_image = X_test[0:1]

prediction = model.predict(test_image, verbose=0)

predicted_class = np.argmax(prediction[0])

print("\nActual Class    :", y_test[0])
print("Predicted Class :", predicted_class)

print("\nCNN Program Executed Successfully!")
```
