```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# 1. Create sample image data
np.random.seed(42)

x = np.random.rand(500, 32, 32, 3).astype("float32")
y = np.random.randint(0, 43, 500)

# 2. Split into training and testing data
x_train = x[:400]
y_train = y[:400]

x_test = x[400:]
y_test = y[400:]

# 3. Create CNN model
model = Sequential([
    Conv2D(32, (3, 3), activation="relu",
           input_shape=(32, 32, 3)),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Flatten(),
    Dense(128, activation="relu"),
    Dense(43, activation="softmax")
])

# 4. Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 5. Train model
history = model.fit(
    x_train,
    y_train,
    epochs=3,
    batch_size=32,
    validation_split=0.2
)

# 6. Evaluate model
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)

print("\nTest Loss:", loss)
print("Test Accuracy:", accuracy)

# 7. Predict one test image
test_image = x_test[0]

# Add batch dimension
test_image_batch = np.expand_dims(test_image, axis=0)

prediction = model.predict(test_image_batch, verbose=0)

# Get predicted class
predicted_class = np.argmax(prediction[0])

print("\nActual Class    :", y_test[0])
print("Predicted Class :", predicted_class)

print("\nCNN program executed successfully!")
```
