```python

import tensorflow as tf
import matplotlib.pyplot as plt

# Load CIFAR-10 dataset
(x, _), _ = tf.keras.datasets.cifar10.load_data()
x = tf.image.resize(x[:1000], (32, 32))
x = (x / 127.5) - 1

# Simple generator
G = tf.keras.Sequential([
    tf.keras.layers.Input((100,)),
    tf.keras.layers.Dense(32*32*3, activation="tanh"),
    tf.keras.layers.Reshape((32, 32, 3))
])

# Generate images
noise = tf.random.normal((8, 100))
images = (G(noise).numpy() + 1) / 2

# Display generated images
plt.figure(figsize=(8, 2))
for i in range(8):
    plt.subplot(1, 8, i+1)
    plt.imshow(images[i])
    plt.axis("off")
plt.show()

print("Generated images successfully.")
```
